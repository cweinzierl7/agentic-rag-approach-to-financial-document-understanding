# Evaluation Visualizations & Statistical Analysis

This notebook analyzes the performance of **single-agent** vs **multi-agent** architectures for credit risk KPI retrieval from financial documents.

## Overview

**Datasets**: RWE and Walmart financial reports  
**Metrics Evaluated**:
- **Agent Goal Accuracy**: Correctness of retrieved KPI values
- **Tool Call F1**: Efficiency and precision of tool usage
- **Combined Score**: Weighted aggregate of accuracy and efficiency

**Analysis Sections**:
1. Qualitative KPI comparison 
2. Quantitative KPI comparison
3. Temporal analysis across fiscal years
4. Derived KPI evaluation (computed metrics)

## Visualizations: Quantitative Performance Comparison

```python
import matplotlib.pyplot as plt

# Get the Blues colormap
blues = plt.cm.Blues

# Get a specific color (value between 0.0 and 1.0)
# 0.0 = lightest, 1.0 = darkest
light_blue = blues(0.3)   # lighter shade
medium_blue = blues(0.5)  # medium shade
dark_blue = blues(0.9)    # darker shade

size_axes = 13 +5
size_subtitle = 14+5
size_title = 16+5
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 6)

# Create figure with subplots for RWE and Walmart
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# RWE Quantitative Comparison
rwe_data = pd.DataFrame({
    'Approach': ['Single Agent', 'Multi-Agent'],
    'Accuracy': [df_rwe_quant_single['agent_goal_accuracy'].mean(), 
                 df_rwe_quant_multi['agent_goal_accuracy'].mean()]
})

# Plot RWE Accuracy
ax1 = axes[0]
bars1 = ax1.bar(rwe_data['Approach'], rwe_data['Accuracy'], color=[ dark_blue, medium_blue])
ax1.set_title('RWE Quantitative: Agent Goal Accuracy', fontsize=14, fontweight='bold')
ax1.set_ylabel('Agent Goal Accuracy', fontsize=12)
ax1.set_ylim(0, 1.0)
ax1.grid(axis='y', alpha=0.3)

# Add value labels on bars
for i, bar in enumerate(bars1):
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.3f}',
             ha='center', va='bottom', fontsize=11, fontweight='bold')

# Walmart Quantitative Comparison
walmart_data = pd.DataFrame({
    'Approach': ['Single Agent', 'Multi-Agent'],
    'Accuracy': [df_walmart_quant_single['agent_goal_accuracy'].mean(), 
                 df_walmart_quant_multi['agent_goal_accuracy'].mean()]
})

# Plot Walmart Accuracy
ax2 = axes[1]
bars2 = ax2.bar(walmart_data['Approach'], walmart_data['Accuracy'], color=[ dark_blue, medium_blue])
ax2.set_title('Walmart Quantitative: Agent Goal Accuracy', fontsize=14, fontweight='bold')
ax2.set_ylabel('Agent Goal Accuracy', fontsize=12)
ax2.set_ylim(0, 1.0)
ax2.grid(axis='y', alpha=0.3)

# Add value labels on bars
for i, bar in enumerate(bars2):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.3f}',
             ha='center', va='bottom', fontsize=11, fontweight='bold')

plt.tight_layout()
plt.savefig('../plots_eval/quantitative_accuracy_comparison.png', dpi=300, bbox_inches='tight')
plt.show()
```
