# Evaluation Visualizations & Statistical Analysis

This notebook analyzes the performance of **single-agent** vs **multi-agent** architectures for credit risk KPI retrieval from financial documents.

## Overview

**Metrics Evaluated**:
- **Agent Goal Accuracy**: Correctness of retrieved KPI values
- **Tool Call F1**: Efficiency and precision of tool usage
- **Combined Score**: Weighted aggregate of accuracy and efficiency

**Analysis Sections**:
1. Qualitative KPI comparison 
2. Quantitative KPI comparison
3. Temporal analysis across fiscal years
4. Derived KPI evaluation (computed metrics)

# Qualitative KPI Analysis

```python
import pandas as pd
import os

# Check current working directory
print(f"Current directory: {os.getcwd()}")

# List files in evaluation folder to verify the path
if os.path.exists(CSV_BASE_PATH):
    print(f"Files in {CSV_BASE_PATH}:")
    print(os.listdir(CSV_BASE_PATH))

# Helper function to load CSV with error handling
def load_csv_safe(company, suffix, base_path=CSV_BASE_PATH):
    """Load CSV file with error handling"""
    filepath = f"{base_path}{company}_{suffix}.csv"
    try:
        df = pd.read_csv(filepath)
        print(f"✓ Loaded: {filepath}")
        return df
    except FileNotFoundError:
        print(f"✗ Not found: {filepath}")
        return None

# Load qualitative data for both companies
df_company1_qual_single = load_csv_safe(COMPANY_1, FILE_SUFFIX_QUAL_SINGLE)
df_company1_qual_multi = load_csv_safe(COMPANY_1, FILE_SUFFIX_QUAL_MULTI)

if COMPANY_2:
    df_company2_qual_single = load_csv_safe(COMPANY_2, FILE_SUFFIX_QUAL_SINGLE)
    df_company2_qual_multi = load_csv_safe(COMPANY_2, FILE_SUFFIX_QUAL_MULTI)
```

```python
# Statistical overview of the dataframe (Company 1 - Single Agent)
print(f"=== {COMPANY_1} Qualitative Single Agent Statistics ===\n")
print("Basic Statistics:")
print(df_company1_qual_single.describe())
print("\n" + "="*80 + "\n")

print("Data Types and Non-Null Counts:")
print(df_company1_qual_single.info())
print("\n" + "="*80 + "\n")

print("Shape (rows, columns):", df_company1_qual_single.shape)
print("\n" + "="*80 + "\n")

print("Agent Goal Accuracy Statistics:")
print(f"Mean: {df_company1_qual_single['agent_goal_accuracy'].mean():.3f}")
print(f"Median: {df_company1_qual_single['agent_goal_accuracy'].median():.3f}")
print(f"Min: {df_company1_qual_single['agent_goal_accuracy'].min():.3f}")
print(f"Max: {df_company1_qual_single['agent_goal_accuracy'].max():.3f}")
print(f"Std Dev: {df_company1_qual_single['agent_goal_accuracy'].std():.3f}")
print("\n" + "="*80 + "\n")

print("Tool Calls Statistics:")
print(f"Mean: {df_company1_qual_single['num_tool_calls'].mean():.2f}")
print(f"Median: {df_company1_qual_single['num_tool_calls'].median():.0f}")
print(f"Min: {df_company1_qual_single['num_tool_calls'].min()}")
print(f"Max: {df_company1_qual_single['num_tool_calls'].max()}")
print("\n" + "="*80 + "\n")

print("Missing Values:")
print(df_company1_qual_single.isnull().sum())
print("\n" + "="*80 + "\n")

print("Unique KPIs:")
print(f"Number of unique KPIs: {df_company1_qual_single['kpi_name'].nunique()}")
```

#### Comparative Analysis: Single Agent vs Multi-Agent

Comparing performance metrics between single-agent and multi-agent approaches for both companies.

```python
# Combined Summary: All Datasets
print("="*100)
print("COMBINED SUMMARY: ALL DATASETS")
print("="*100)

# Build summary dynamically based on available data
datasets = []
avg_accuracy = []
median_accuracy = []
avg_tool_calls = []
max_tool_calls = []
num_kpis = []

# Company 1 data
if df_company1_qual_single is not None:
    datasets.append(f'{COMPANY_1} Qual (Single)')
    avg_accuracy.append(df_company1_qual_single['agent_goal_accuracy'].mean())
    median_accuracy.append(df_company1_qual_single['agent_goal_accuracy'].median())
    avg_tool_calls.append(df_company1_qual_single['num_tool_calls'].mean())
    max_tool_calls.append(df_company1_qual_single['num_tool_calls'].max())
    num_kpis.append(df_company1_qual_single['kpi_name'].nunique())

if df_company1_qual_multi is not None:
    datasets.append(f'{COMPANY_1} Qual (Multi)')
    avg_accuracy.append(df_company1_qual_multi['agent_goal_accuracy'].mean())
    median_accuracy.append(df_company1_qual_multi['agent_goal_accuracy'].median())
    avg_tool_calls.append(df_company1_qual_multi['num_tool_calls'].mean())
    max_tool_calls.append(df_company1_qual_multi['num_tool_calls'].max())
    num_kpis.append(df_company1_qual_multi['kpi_name'].nunique())

# Company 2 data (if exists)
if COMPANY_2 and df_company2_qual_single is not None:
    datasets.append(f'{COMPANY_2} Qual (Single)')
    avg_accuracy.append(df_company2_qual_single['agent_goal_accuracy'].mean())
    median_accuracy.append(df_company2_qual_single['agent_goal_accuracy'].median())
    avg_tool_calls.append(df_company2_qual_single['num_tool_calls'].mean())
    max_tool_calls.append(df_company2_qual_single['num_tool_calls'].max())
    num_kpis.append(df_company2_qual_single['kpi_name'].nunique())

if COMPANY_2 and df_company2_qual_multi is not None:
    datasets.append(f'{COMPANY_2} Qual (Multi)')
    avg_accuracy.append(df_company2_qual_multi['agent_goal_accuracy'].mean())
    median_accuracy.append(df_company2_qual_multi['agent_goal_accuracy'].median())
    avg_tool_calls.append(df_company2_qual_multi['num_tool_calls'].mean())
    max_tool_calls.append(df_company2_qual_multi['num_tool_calls'].max())
    num_kpis.append(df_company2_qual_multi['kpi_name'].nunique())

summary_all = pd.DataFrame({
    'Dataset': datasets,
    'Avg Accuracy': avg_accuracy,
    'Median Accuracy': median_accuracy,
    'Avg Tool Calls': avg_tool_calls,
    'Max Tool Calls': max_tool_calls,
    'Num KPIs': num_kpis
})

print("\n")
print(summary_all.to_string(index=False))
print("\n" + "="*100 + "\n")
```
